```python
import sys
from pathlib import Path
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import fsolve, minimize

sys.path.append('.')
from src.fluid import Fluid
from src.reservoir import ResProps, Reservoir
from src.pipe import Pipe
from src.well import Well
from src.compressor import DCS
from src.simulator import FieldSimulator

print("Библиотеки импортированы")
```

```python
BASE = Path(".")
fluid = Fluid(interp_csv_path=str(BASE / "interp_data.csv"))
print(f"Fluid создан: T = {fluid.T} K, M = {fluid.M} кг/моль")
```

```python
P_range = np.linspace(1, 200, 200)
Z_vals = [fluid.z(p) for p in P_range]
Bg_vals = [fluid.bg(p) for p in P_range]
mu_vals = [fluid.mu(p) for p in P_range]
rho_real = [fluid.ro(p) for p in P_range]
rho_ideal = [fluid.ro_ideal(p) for p in P_range]

fig, axs = plt.subplots(2, 2, figsize=(12, 10))
axs[0,0].plot(P_range, Z_vals); axs[0,0].set_xlabel('P, атм'); axs[0,0].set_ylabel('Z'); axs[0,0].set_title('Z-фактор'); axs[0,0].grid()
axs[0,1].plot(P_range, Bg_vals); axs[0,1].set_xlabel('P, атм'); axs[0,1].set_ylabel('Bg, м³/м³'); axs[0,1].set_title('Объёмный фактор'); axs[0,1].grid()
axs[1,0].plot(P_range, mu_vals); axs[1,0].set_xlabel('P, атм'); axs[1,0].set_ylabel('μ, сП'); axs[1,0].set_title('Вязкость'); axs[1,0].grid()
axs[1,1].plot(P_range, rho_real, label='Реальный')
axs[1,1].plot(P_range, rho_ideal, '--', label='Идеальный')
axs[1,1].set_xlabel('P, атм'); axs[1,1].set_ylabel('ρ, кг/м³'); axs[1,1].legend(); axs[1,1].set_title('Плотность'); axs[1,1].grid()
plt.tight_layout()
plt.show()
```

```python
V_res = math.pi * 500**2 * 25
resprops = ResProps(P=100.0, V=V_res, T=fluid.T)
reservoir = Reservoir(resprops, fluid)

wells = [
    Well(fluid, k=50, h=25, re=500, rw=0.1, pipe=Pipe(2000, 0.062, 0.000046, fluid, vertical_depth=1800)),
    Well(fluid, k=50, h=25, re=500, rw=0.1, pipe=Pipe(2500, 0.062, 0.000046, fluid, vertical_depth=1900)),
    Well(fluid, k=50, h=25, re=500, rw=0.1, pipe=Pipe(1800, 0.073, 0.000046, fluid, vertical_depth=1600)),
]
print(f"Создано {len(wells)} скважин")
```

```python
P_res_fixed = 100.0
P_bhp_range = np.linspace(10, 100, 50)
q_ipr = [wells[0].q(P_res_fixed, pb) for pb in P_bhp_range]

plt.figure(figsize=(8,5))
plt.plot(q_ipr, P_bhp_range)
plt.xlabel('q, ст.м³/сут')
plt.ylabel('P_bhp, атм')
plt.title('IPR скважины 1 при P_res = 100 атм')
plt.gca().invert_yaxis()
plt.grid()
plt.show()
```

```python
D_test = 0.062
eps_test = 0.000046
Re_range = np.logspace(4, 7, 100)
lam_vals = []
for Re in Re_range:
    if Re < 2300:
        lam = 64/Re
    else:
        def colebrook(lam):
            return 1/np.sqrt(lam) + 2*np.log10(eps_test/(3.7*D_test) + 2.51/(Re*np.sqrt(lam)))
        lam = fsolve(colebrook, 0.02)[0]
    lam_vals.append(lam)

plt.figure(figsize=(8,5))
plt.loglog(Re_range, lam_vals)
plt.xlabel('Re'); plt.ylabel('λ'); plt.title('Коэффициент трения (Колбрук-Уайт)')
plt.grid()
plt.show()
```

```python
P_wh_fixed = 10.0
q_range = np.linspace(0, 2000, 50)
plt.figure(figsize=(10,6))
for i, well in enumerate(wells):
    P_bhp_vals = [well.vlp(P_wh_fixed, q) for q in q_range]
    plt.plot(q_range, P_bhp_vals, label=f'Скважина {i+1}')
plt.xlabel('q, ст.м³/сут')
plt.ylabel('P_bhp, атм')
plt.title('VLP при P_wh = 10 атм')
plt.legend()
plt.grid()
plt.gca().invert_yaxis()
plt.show()
```

```python
shlyf = Pipe(L=5000, D=0.200, roughness=0.000046, fluid=fluid, vertical_depth=0.0)
dcs = DCS(CR=1.5, P_line=5.0, q_ext=500.0)
sim = FieldSimulator(reservoir, wells, shlyf, dcs)
states = sim.solve(P_res=100.0)

print("Рабочая точка при P_res = 100 атм:")
for name, st in states.items():
    print(f"{name}: q={st.q_std:.1f} ст.м³/сут, P_in={st.P_in:.2f} атм, P_out={st.P_out:.2f} атм")
```

```python
P_man = states['shlyf'].P_in
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for i, ax in enumerate(axes):
    well = wells[i]
    q_vals = np.linspace(0, 2000, 100)
    # IPR: P_bhp = P_res - q / C
    mu = fluid.mu(100.0)
    C = (well.k * well.h) / (mu * np.log(well.re/well.rw)) * 0.01   # см. well.productivity_index
    P_bhp_ipr = [100.0 - q / C for q in q_vals]
    P_bhp_vlp = [well.vlp(P_man, q) for q in q_vals]
    ax.plot(q_vals, P_bhp_ipr, label='IPR')
    ax.plot(q_vals, P_bhp_vlp, label='VLP')
    q_work = states[f'well_{i+1}'].q_std
    P_work = well.vlp(P_man, q_work)
    ax.plot(q_work, P_work, 'ro', label='Рабочая точка')
    ax.set_xlabel('q, ст.м³/сут')
    ax.set_ylabel('P_bhp, атм')
    ax.set_title(f'Скважина {i+1}')
    ax.legend()
    ax.grid()
    ax.invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
reservoir.resprops.P = 100.0
reservoir.Gp = 0.0
df = sim.run(N_days=180, dt=1.0)
df.head()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

# Функция безопасного преобразования в float
def safe_float(x):
    if x is None:
        return 0.0
    if isinstance(x, (np.ndarray, list, tuple)):
        if len(x) == 0:
            return 0.0
        x = x[0]
    try:
        return float(x)
    except (TypeError, ValueError):
        return 0.0

try:
    # Загрузка данных
    test = pd.read_csv('test.csv', sep=',')
    if 'p' not in test.columns or 'qgas' not in test.columns:
        raise KeyError("В test.csv нет столбцов 'p' или 'qgas'")
    
    q_fact = test['qgas'].values.astype(float)
    P_res_hist = test['p'].values.astype(float)
    
    # Сохраняем исходные проницаемости
    original_ks = [w.k for w in wells]
    
    # Целевая функция
    def objective(C_mult):
        for w in wells:
            w.k = 50 * C_mult
        sim.wells = wells
        q_sim = []
        for p in P_res_hist:
            st = sim.solve(p)
            total = 0.0
            for i in range(1, 4):
                val = st[f'well_{i}'].q_std
                total += safe_float(val)
            q_sim.append(total)
        # Преобразуем в одномерный массив float
        q_sim = np.array(q_sim, dtype=float).flatten()
        return np.sqrt(np.mean((q_sim - q_fact)**2))
    
    # Оптимизация
    res = minimize(objective, 1.0, method='Nelder-Mead', options={'maxiter': 200})
    C_opt = res.x[0]
    print(f"Оптимальный множитель проницаемости: {C_opt:.3f}")
    
    # Восстановление исходных k
    for w, k_orig in zip(wells, original_ks):
        w.k = k_orig
    
    # Применение оптимального множителя
    for w in wells:
        w.k = 50 * C_opt
    sim.wells = wells
    
    # Расчёт модельных дебитов после калибровки
    q_sim_calib = []
    for p in P_res_hist:
        st = sim.solve(p)
        total = 0.0
        for i in range(1, 4):
            val = st[f'well_{i}'].q_std
            total += safe_float(val)
        q_sim_calib.append(total)
    q_sim_calib = np.array(q_sim_calib, dtype=float).flatten()
    q_fact = np.array(q_fact, dtype=float).flatten()
    
    # Метрики
    ss_res = np.sum((q_sim_calib - q_fact)**2)
    ss_tot = np.sum((q_fact - np.mean(q_fact))**2)
    R2 = 1 - ss_res / ss_tot if ss_tot != 0 else 0.0
    RMSE = np.sqrt(np.mean((q_sim_calib - q_fact)**2))
    print(f"R² = {R2:.4f}, RMSE = {RMSE:.2f} ст.м³/сут")
    
    # График
    plt.figure(figsize=(8,6))
    plt.scatter(q_fact, q_sim_calib, alpha=0.7)
    min_val = min(q_fact.min(), q_sim_calib.min())
    max_val = max(q_fact.max(), q_sim_calib.max())
    plt.plot([min_val, max_val], [min_val, max_val], 'r--', lw=2)
    plt.xlabel('Фактический дебит (test.csv), ст.м³/сут')
    plt.ylabel('Модельный дебит, ст.м³/сут')
    plt.title('Калибровка по данным test.csv')
    plt.grid(True)
    plt.show()

except FileNotFoundError:
    print("Файл test.csv не найден. Калибровка пропущена.")
except Exception as e:
    print(f"Ошибка при калибровке: {e}")
```
